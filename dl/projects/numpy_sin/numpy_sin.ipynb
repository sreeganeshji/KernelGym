```python
import numpy as np
import torch
from torch import Tensor

```

```python
torch.manual_seed(42)

a = torch.randn(3,2)
a_pre = a.storage();
a_t = a.T
print(f"{a.storage()}")
torch.testing.assert_close(a_t.storage(), a.storage())
# torch.testing.assert_close(a_t.stride(), a.stride())
print(f"{a} and \n {a_t}")
print(f"{a.stride()} and {a_t.stride()}")
b = Tensor.contiguous(a_t)
print(f"{b}, \n {b.shape} \n {b.stride()}")
print(f"{b.storage()}")
print(f"{a.is_contiguous()} {a_t.is_contiguous()} {b.is_contiguous()}")
```

```python
#Chunking!!
'''
Say if you create a new tensor with just the row of a[1] which is a 3x2 tensor, 
then the new tensor b will actually still be pointing to the a's memory but just with an offset!
'''
torch.manual_seed(42)
a = torch.randn((4,3,2))
b = a[1]
print(f"{a.shape} {b.shape}")
print(f"{a.storage_offset()} {b.storage_offset()}")
for (a_s, b_s) in zip(a.storage(), b.storage()):
    if(a_s != b_s):
        print("Wrong!")
```
