```python
import torch
import torch.nn as nn

torch.manual_seed(123)
x = torch.randint(0, 10, [10,1])
N = x.shape[0]
rmsNorm = nn.modules.RMSNorm([10,1], 0.0)
rms = rmsNorm(x.float())
rms
```

```python

RMS = x**2 
RMS = torch.sum(RMS, dim=0)
RMS = RMS / N
RMS = torch.sqrt(RMS)
x/RMS
```
